## 4. Experimente: mude a razão de projeto r

Altere o valor de `r` e rode a célula de novo. Valores maiores deixam o piloto mais rápido e
fazem o leme trabalhar mais; valores menores deixam o ajuste mais suave.

```python
import simulacao_prova_de_conceito as sim
%matplotlib inline
import matplotlib.pyplot as plt

r = 2.5   # razão de projeto: mude aqui
psi, dc, d = sim.simular(sim.sintonia_tabela1(r))
m = sim.indicadores(psi, dc, d)

fig, ax = plt.subplots(figsize=(8, 3.5))
ax.axvspan(sim.PERT_INI, sim.PERT_FIM, color="#f0efec", lw=0, label="Perturbação")
ax.plot(sim.t, sim.ref, "--", color="gray", label="Proa pedida")
ax.plot(sim.t, psi, label=f"Świder, Trybus e Stec, r = {r}")
ax.set_xlabel("Tempo (s)"); ax.set_ylabel("Proa (graus)"); ax.legend(loc="lower right")
plt.show()
print(f"IAE = {m['iae']:.1f} grau·s | maior desvio sob perturbação = {m['desvio']:.2f}° | "
      f"atividade do leme = {m['atividade']:.2f}°")
```
